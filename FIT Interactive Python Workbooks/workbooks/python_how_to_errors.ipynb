# How to work with and understand Errors

It's every programmers goal to write exceptional code, but unfortunately some developers write different types of exceptional code... Within programming there are three categories of errors or exceptions. (depending on the language you are using, they may be called differently) 

- Syntactical Errors

- Runtime Errors

- Logical Errors


# Syntactical Errors

Syntax errors are going to quickly become your favourite errors. This is simply because theye are the easiest to fix, and tend not to cause major _(depending on your opinion of major, as they can create logical errors inadvertently!)_ problems. Plus the more you write code, the less of these you will tend to make. They are simply mistakes with how your code is written - when it comes to fixing them, most languages will point you to the exact line where the error is, what more could you ask for? We're also spoiled by IDE (internal development environments) that have prompts and checks for us to catch the simple stuff.

Most syntax errors tend to be made when you're new to or haven't used a particular language in a while. Things like missing semi-colons, incorrect brackets, improper type conversion and missing curly brackets (Java/C# programmers!). Basically a wide variety of small problems that can be easily fixed.


### Spot The Syntax Error:

> if 5 < 5;
>
>       print("Five is less than five!")

The error is of course the fact that the programmer used a semi-colon on the if construct rather than the required colon. This is a very simple problem to fix as our code won't even run in it's current state. In-fact when we try to run it, we even get a very helpful Error output pointing us directly to the line of code that is causing the problem.

# Runtime Errors

Runtime errors can be caused by a variety of things. From incorrect values being passed to in-accessible databases, servers or files. These are errors that we can account for and build redundancies into our code to handle such things occurring.

Basically our program will be truckin' along performing all of it's operations and actions until something unexpected occurs. Whilst me and you with our human brain can figure out how to solve an unexpected problem a computer obviously can not, unless instructed to. 

These errors tend to be things outside of our control. For example, when we release our program into the wild that is interacting with a database, or storing information into a file. If the user were to remove the config file or the database server were to go down, there's not much we can do to fix that. What we can do though is essentially build a back-up plan into our code. Instruct the program what to do when issues like this occur. 

Remember that sometimes it's impossible to do anything if a runtime error occurs, all you might be able to do is log the error - display an error message and close the program gracefully.

# Logical Errors

Logical errors are arguably the worst kind of errors that we can come across within programming. This is because they tend to be mistakes made by a programmer in the logic of code. They may not present themselves in any form of output or red squiggly line, but can cause havoc to the program by making it behave not as intended, depending on the severity of the error. 

These are the kind of errors that are often referred to as bugs, and as you've probably heard before bug hunting can take teams of people lots of time.

We can of course simply call the bug a feature and move onto something more exciting... right?

```python
# Below is a very basic example of a logical error. Can you spot it?

numbers = [1, 2, 3, 4, 5]
sum_result = 0

for num in numbers:
    sum_result += num

print(f"The sum of numbers is: {sum_result}")
```
