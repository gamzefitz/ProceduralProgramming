## Taking in input

Before we look at some of the constructs available to us, firstly we'll take a look at how to retrieve some input from the user. This will make things a little easier when it comes to writing some code.

Taking in input in Python is very easy thanks to one of the built-in functions called input() 

You can find out more about the built-in functions here: https://docs.python.org/3/library/functions.html

    input() will allow you to provide a prompt to the user and retrieve some input in the form of a string

Below you can see an example of the input function in action. As noted above, the function returns a string so if we wanted to convert the string to an int we would need to use the int() function

```python
user_name = input("What is your name? ")
print(f"Output: {user_name}")

user_age = int(input("What is your age? ")) # Note this can cause an error if the user doesn't enter a number, more on error handling later
print(f"Output: {user_age}")

user_python_pref = input("Do you like Python? (Y/N) ")
print(f"You said {user_python_pref} when asked if you liked Python")
```

# Conditional Statements

Conditional statements are used to essentially ask a true or false question and perform an operation based on the answer.

## The if Statement

The if statement is the bread and butter of conditional statements, as programmers we use it to control the flow of a program, essentially giving us the ability to make choices within our applications. 

The syntax for an if statement is relatively straight forward - the main thing you'll notice coming from different languages is the distinct lack of brackets and code blocks. This down to the white spaces being so important.

    if condition : 
        statement that executes if the condition results in true

The first portion of the statement is the if keyword. Next is the conditional portion of the statement, this is where we define the code that will result in a boolean value, true or false. We denote the end of the conditional statement using a colon : so our conditional statement is essentially wrapped between the if and the colon.

As stated above the first difference you might notice depending on the language you come from is the lack of a code block afterwards. This is where the whitespaces come into play. What belongs to the if statement is marked by indenting it on the following lines of code.


```python
bool_sample = True

if bool_sample :
    print("Hello")
    
bool_sample = not bool_sample

if bool_sample :
    print("World")
```

You can see in the above snippet of code that only "Hello" was printed out despite us having a print for "Hello" also. This is because the condition, which is this case was just simply a boolean, was false on the second if statement. This meant that the print statement did not execute.

## The elif keyword

There are going to be many situations where you want to check for more than just one condition and perform different operations depending on which condition is true. This is where the elif keyword comes into play.

    if condition :
        print("Hello World")
    elif condition :
        print("Goodbye World")
    elif condition : 
        print("Maybe World")
        
The above demonstrates the syntax of including multiple elif statements. The important thing to keep in mind is that these all form one structure, meaning that only one of the indented statements will execute. The code will check each of them from top to bottom and stop checking after it hits the first one which is true.


```python
# What will be output by the below code?

if 5 >= 7 :
    say = "nay"
elif 5 <= 7 :
    say = "ni"
elif 5 != 7 :
    say = "noi"

print("The knights who say " + say)
```

## The else keyword

The last thing we're going to add to the if construct is the else keyword. This essentially acts as a catch-all for the entire construct. This means that if none of the if or elif statements execute, the else will execute.

    if condition :
        print("Hello")
    elif condition :
        print("There.")
    elif condition :
        print("General")
    else :
        print("Kenobi!")

As you can see the structure of the else statement is very simple, we do not specify a condition as it acts as a 'catch-all' when none of the above options execute.
