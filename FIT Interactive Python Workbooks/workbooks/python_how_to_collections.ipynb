# How to work with collections in Python

Collections are containers that allow you to group multiple elements into a single entity. They provide a convenient way to organise and manipulate data. There are many different types of collections, such as lists as tuples, each with their own characteristics and use cases. It's important to understand the features of a collection and any potential drawbacks before opting to implement them within your application.

This workbook has been designed to provide an introduction to working with two types of collections within Python, lists and tuples. The goal is to provide a working knowledge of both to allow you to get started with implementing collections. As is always the case, it is recommended to dive deeper into the documentation, official resources and practice to become familiar with collections.


# How to work with lists in Python. 

A list is a collection of items that are ordered and changeable. In this example, we create a list of fruits and perform various operations on it.

First, we access elements of the list using their index. We can also slice a list to get a subset of its elements. We can modify elements of a list, add new elements to it, and remove elements from it. We can also check if an element is in a list and get the length of a list.

```python
# Create a list of fruits
fruits = ['apple', 'banana', 'cherry']

# Accessing elements of a list
print('Accessing elements of a list:')
print(fruits[0])  # Output: 'apple'
print(fruits[1])  # Output: 'banana'
print(fruits[-1])  # Output: 'cherry'

# Getting the length of a list using the built-in len() function
print(f'\nLength of the fruits list is: {len(fruits)}')
```

#### Adding to 'append()' and removing from 'remove()' a list

Once we have created our list of fruits in the cell above, we can see how to both add to, using the append function and remove from, using the remove function

```python
# Adding elements to a list
fruits.append('grape')
print(fruits)  # Output: ['apple', 'banana', 'cherry', 'grape']

# Removing elements from a list
fruits.remove('banana')
print(fruits)  # Output: ['apple', 'cherry', 'grape']

# Modifying elements of a list
fruits[1] = 'orange'
print(fruits)  # Output: ['apple', 'orange', 'grape']
```

We have to be careful when removing from a list using the 'remove()' function, as it can cause an exception when an element is not present

Try to run the code below and see what happens!

```python
# Removing an element from a list that does not exist
fruits.remove('tomato')
```

#### Checking if an element is in the list

We can use the in keyword to check if a particular element exists in a list (or any iterable like tuple, set, dictionary, string etc.). It will return a boolean value of True if the element is found in the list, and False otherwise. 

```python
# Checking if an element is in a list
print('apple' in fruits)  # Output: True
print('mango' in fruits)  # Output: False
```

#### Slicing lists

Python has some built-in features that allow us to access part of sequences (like strings, tuples, and lists). We can slice a list by specifying a start (inclusive) and end index (exclusive) giving us a copy of the respective elements from the list 

```python
# Slicing a list

# Outputs assuming fruits = ['apple', 'orange', 'grape']

print(fruits[1:3])  # Output: ['orange', 'grape']
print(fruits[:2])  # Output: ['apple', 'orange']
print(fruits[1:])  # Output: ['orange', 'grape']
```

# How to work with Tuples in Python

A tuple is an ordered, immutable collection of elements. It is similar to a list, but unlike lists it cannot be modified once they are created. The syntax for defining a tuple uses parentheses '()'  rather than square brackets that we're used to from lists, this enables us to differentiate them simply.

> __Reminder:__ Immutability / Immutable - This ensures that once a variable is assigned a value, that value cannot be chnged or modified. This provides us with stability and predictability in our code.

## Creating Tuples

Creating tuples is easy, we simply declare a variable name to hold our new tuple and assign it a value of () this will create an empty tuple. We can add some elements to our tuple by providing comma separated values between the () that we declared. See below.

Tuples in Python are very versatile and can hold a variety of values, including Numbers, Strings, Booleans, Other Tuples, and even a mix of data types. This flexibility makes tuples useful in many different scenarios, such as representing coordinates, storing configurations, or returning mutliple values from a function.

```python
# Creating tuples

empty_tuple = ()

fruit_tuple = ('apple', 'banana', 'cherry')

number_tuple = (1, 2, 3)

coordinates_tuple = ((3,5),(1, 2))
```

## Accessing Elements

We can access elements in a tuple using indexing, again similar to our lists.

```python
# Accessing elements of a tuple

first_fruit = fruit_tuple[0]
print(first_fruit)  # Output: 'apple'

second_fruit = fruit_tuple[1]
print(second_fruit)  # Output: 'banana'

# Negative indexing
last_fruit = fruit_tuple[-1]
print(last_fruit)  # Output: 'cherry'
```

## Immutability

It is important that we fully understand the immutable nature of tuples before trying to implement them within our code. Given how we understand immutable objects within code, can you understand why the below code would throw an error?

```python
# The following line of code will throw an error
fruit_tuple[0] = 'mango'
```

You may end up asking the purpose of something like an empty tuple, given the immutable nature of them, the below demonstrates how we can start with an empty tuple but add elements at a later stage. This can arise in situations where you need to start with an empty tuple and accumulate data over-time or even just have a placeholder that may end up remaining empty.

```python
data = () # Empty tuple

# Some processing
data += ('new data',)

# More processing
data += ('more data',)
```

This can be a useful feature or approach for things like return values from functions (more on functions later on!)

## Tuple Methods

Tuples unfortunately lack many of the methods available for lists, however they do still have a few similar ones for utilising, such as;

- __count__: Return number of occurrences of value.
- __index__: Return first index of value.Raises ValueError if the value is not present. 

```python
number_of_apples = fruit_tuple.count('apple')

index_cherry = fruit_tuple.index('cherry')
```
