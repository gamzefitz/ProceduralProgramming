Python has a strong emphasis on code readability with whitespaces playing a bigger role than people may expect.

What all of this means is that Python is a very easy to learn language which aims to assist programmers in writing clean, clear and logical code.

### - High Level

# Variables & Simple Data Types

As programmers, we regularly need to declare, manipulate, and interact with values which are stored in memory, this is done using what we call variables. Essentially variables allow us to store and manage data within a program. 

Variables play a crucial role in all programming languages, not just Python. Understanding how to use variables is fundamental to writing effective and readable code. 

## Declaring and Assigning Variables

As discussed above, Python is a dynamically typed language. You'll see in the below example we simply decide upon a variable name and write our code around it. The interpreter will sort out types for us at runtime.

```python
# Declare and assign a value to a variable
message = "Hello Python world!"

# Print the value of the variable
print(message)
```

Assigning variables is done using the = operator. Once we assign a value to a variable, we are essentially storing that value into memory and using the variable as a way to 'point to' that space in memory. 

```python
x = 10 

name = "John"
```

If you're coming to python from a statically typed language, you may notice we haven't specified what type of variable x or name is above. This is because Python is what's known as a _dynamically_ typed language.

As Python is a dynamically typed language, we do not need to declare the data type of a variable explicility. Instead, the interpreter determines the type based on the assigned value. 

```python
x = 10 # x is an integer

name = "John"  # name is a string

# We can quickly check the type by using the type() function to see what type Python assigned to each variable
print(type(x))
print(type(name))
```

### Mathematical Operators
- \+     Addition - returns the sum of both operands
- \-     Subtraction - returns the result of minusing the right operand from the left operand
- \*     Multiplication - returns the result of the operands times by eachother
- /     Division - returns the result of dividing the left operand by the right operand

## Slicing Functionality

A great feature available to us within the Python language is the ability to slice. This will work for strings as well as other constructs that we'll see later such as lists. It gives us the ability to perform more elaborate operations using the index positions


```python
my_slicer = "Liverpool FC"

sliced = my_slicer[0:9]
print("Slice from 0:9 -> " + sliced)

sliced = my_slicer[10:]
print("Slice from 10: -> " + sliced)
```

## Slicing Elaborated

The way that slicing works is relatively simple, but it can get more complex. As a feature, it can be very powerful and useful to be aware of. Below we'll take a look at some of the syntax of slicing, it is important to take your time and try out each type of slice in the allotted code areas. 

If we take the following snippet of code as the example explained:

```python
slice_example = "The Quick Brown Fox"

"""

Given the above string, we have an index range of 0 to 18 (0:18) or 19 characters in total

The basic syntax for slicing is target[start:stop]

 - start is the index position to start the slice from (inclusive)
 - stop is the index position to stop the slice at (exclusive)

If we want to slice the word Quick from the above string, we would do the following
 - start at index position 4 (Q)
 - stop at index position 9 (k) (remember this is exclusive, so 9 actually refers to the space after the k)
"""

print(slice_example[4:9]) # Outputs:    Quick
print(slice_example[0:3]) # Outputs:    The
print(slice_example[16:19]) # Outputs:  Fox
print(slice_example[10:15]) # Outputs:  Brown

# Note that spaces are included in the index count
```

An important thing to remember is that by adjusting/manipulating the syntax of features can allow us to do more than we might first think. This is the case with many features of programming and why it's so important for us to practice our 'thinking outside of the box' when learning and becoming familiar with new concepts.

With the above in mind, some funky stuff we can do by ommitting the start or the stop parameters include the following;

```python
# If we leave the start index position blank, Python will start at the beginning of the string

print(slice_example[:3]) # Outputs: The
print(slice_example[:9]) # Outputs: The Quick

# If we leave the stop index position blank, Python will stop at the end of the string

print(slice_example[10:]) # Outputs:    Brown Fox
print(slice_example[16:19]) # Outputs:  Fox  
```

Additionally, there is a third parameter we can add to our slicing commands, this is known as the stride. The stride is the number of characters we wis hto move forward, after the first character is retrieved. This means that if we have a stride of 2, we will get every other character.

```python
"""
Including the stride parameter, our syntax for slicing becomes target[start:stop:stride]
    
 - stride is the number of characters to step over each time, the default value is 1
 - stride can be a negative number, which will cause the slice to be performed backwards

 
"""

print(slice_example[:9:2]) # Outputs: TeQik (notice it skips every other character from the start to the stop, including spaces)
print(slice_example[4:9:2]) # Outputs: Qik

# If we want to reverse a string, we can use a stride of -1	
print(slice_example[::-1]) # Outputs: xoF nworB kciuQ ehT (this can act as a quick reverse string function)

# Note we can also use negative values in our start and stop positions, however this is less intuitive and can be confusing
print(slice_example[-3:]) # Outputs: Fox
```

## Relational Operators

There are quite a few relational operators available to us within the language. Most of which you'll be familiar with.

- == - the equality operator returns true if both operands are equal
- < - the less than operator returns true if the operand on the left is less than the operand on the right  
- <= - the less than or equal operator returns true if the operand on the left is less than the operand on the right OR both operands are equal
- \> - the greater than operator returns true if the operand on the left is greater than the operand on the right
- \>= - the greater than or equal operator retursn true if the operand on the left is greater than the operand on the right OR both operands are equal
- != - the not equal to operator returns true if both operands are not equal

```python
# Play around with the relational operators that you feel will you need more practice with


print("5 < 5 is " + str(5 < 5))
print("5 <= 5 is " + str(5 <= 5))
print("5 > 5 is " + str(5 > 5))
print("5 >= 5 is " + str(5 >= 5))
print("5 == 8 is " + str(5 == 8))
print("5 != 8 is " + str(5 != 8))
```

## Logical Operators

There are a number of logical operators that will already be familiar, however, the syntax on them can be a bit different than what you're used to depending on the language you're coming from.

- and - returns true if both the left and the right operand equate to true
- or  - returns true if either the left or the right operand are true
- not - a way of flipping or reversing a boolean value
    
There is also the bitwise version of the logical operators

- &   - returns true if both the left and the right operand equate to true
- |   - returns true if either the left or the right operand are true

```python
# Write some sample code to demonstrate the use of and, or, not, & and | (bitwise operators)

# and
print(True and True) # True
print(True and False) # False
print(False and True) # False
print(False and False) # False

# or
print(True or True) # True
print(True or False) # True
print(False or True) # True
print(False or False) # False

# not
print(not True) # False
print(not False) # True

# &
print(True & True) # True
print(True & False) # False
print(False & True) # False
print(False & False) # False

# |
print(True | True) # True
print(True | False) # True
print(False | True) # True
print(False | False) # False
```

The above code demonstrates logic tables, which are important to be aware of as when you become more proficient with code you'll get to create more advanced compound statements that consist of multiple boolean expressions.
