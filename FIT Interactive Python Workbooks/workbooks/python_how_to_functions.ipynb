# How to work with functions in Python

Functions quickly become the backbone of programming. They empower a programmer to do what all programmers should strive to do, create re-usable snippets of code that can be called upon (invoked) when needed.

# Functions

Functions are somewhat of a back bone to programming. It allows us to do what we should all strive toward doing, creating re-usable snippets of code that can be invoked when needed. 

Coming from a stricter language like Java or C# will make you appreciate the simple approach to functions that Python implements.

## Function Syntax

The syntax for a function in Python is very simple, we use a single keyword __def__ followed by a function name then parameters and the famous __:__ afterwards

```python
def my_first_function():
    print("Wow look at this!")
    print("Such simple syntax!")
```

We can see from the above snippet of code that a function is very simple to write, but don't worry, there's plenty more that we can add. 

## Indentation matters

If we focus on the above example a little more, you'll notice that the print statements are indented below the function definition. This indentation is key to specifying which code actually belongs to the function.

```python
def my_second_function():
    print("This is a print within a function")
print("This is a print NOT within a function")
```

You can see in the above example that there are also two print statements, one however is not indented to be part of the function. This means if we invoke my_second_function() it will only execute one of the print statements. 

## Invoking Functions

Invoking functions is exactly as you might already expect it, once we have the function within scope we simply call it by name followed by the parameters that it requested (more on this later) and it will execute.

```python
# Ensure that you run the above cell which has the functions defined
my_first_function()
```

```python
# Ensure that you run the above cell which has the functions defined
my_second_function()
```

## Function Parameters

A ___lot___ of our functions may require some data in order to perform their operations. We can pass our functions this data via parameters. The parameters that a function requires are set at the same point where we define our function.

The parameters are set in what can be referred to as the parameter list, the set of brackets after the function name. Here we just list the names of the parameters that our function will need in order to operate.

The biggest battle can be appropriately naming the parameters. It is vital that we always use unambiguous and meaningful names so that others reading our code can understand what is happening.

```python
def my_paramed_function(message):
    print(f"The message is '{message}'")
```

The snippet above shows the syntax for specifying parameters. It's clear to see that the above function requires a message in order to execute. Now keep in mind the method name is also extremely important when it comes to naming things clearly and without ambiguity - for the snippets I am providing they are of course not following appropriate convention on the names currently.

## Invoking Parameterized Functions

Invoking parameterized functions is straight forward enough, we just need to ensure that the parameter list matches that of the function's definition.

```python
# Ensure that you run the above cell which has the functions defined
my_paramed_function("The Quick Brown Fox")
```

As you can see, the value that we placed as a parameter for my_paramed_function was then used within the function and referred to as message. 

This provides us with the very important functionality of passing values/data to functions.

```python
# Why won't the below snippet of code work?
my_paramed_function()
```

Remember that when writing functions you're essentially setting out a contract that must be abided by in terms of the parameters that ___must___ be passed to the function.

Otherwise the result is the same as a bove, you simply get an error.

## Multiple Parameters

We can of course specify multiple parameters within the parameter list of a function. The only thing that might get a bit funky coming from a more strictly typed language is ordering your parameters only by name and not type. Further stressing the need for appropriate naming conventions to be followed.

```python
def count_num_letters(sentence, letter, case_sensitive):
    counter = 0
    if not case_sensitive:
        sentence = sentence.lower()
        letter = letter.lower()
    for char in sentence:
        if char == letter:
            counter += 1
    print(f"There are {counter} occurences of '{letter}' in {sentence}")
```

Above is a very simple method used to count the number of a given letter within a given sentence or word. It also allows the ability to specify whether or not case matters in the count.

```python
count_num_letters("banana", "A", False)
```


## Default Value Parameters

Another way we can tweak our parameters is by defining default values for our parameters. Currently the parameters we have defined above have no default values - this means that a value has to be supplied for each parameter when invoking the function. 

What default values allow is for people to not bother passing values for some of the parameters and instead let the function just assume that if they don't specify anything, they must have meant the default value.

```python
def print_number_of_times(msg, num_times=1):
    for i in range(num_times):
        print(msg)
```

```python
print_number_of_times("Spaghetti") # Will print once, as no value was passed for num_times

print_number_of_times("Bolognese", 4) # Will print four times, num_times was given a value
```

The real power is the fact that you don't need to provide values for parameters that can be very easily inferred or set to the norm, only certain times then you have the option to specify a value

# Docstrings

Docstrings are the final thing we're going to take a look at in relation to our functions. These are just how we provide documenation about our functions for those that may be in need of reading it.

We set the doc strings by simply including a multi-line string as the first statement within the function.

```python
def my_undocumented_function():
    print("Good luck figuring me out!")
```

```python
my_undocumented_function.__doc__ # Outputs nothing because there is no doc string
```

```python
def my_documented_function():
    """This is the doc string, here you can read all about my functionality and parameters"""
    print("You can access my doc string by using .__doc__ on my name")
```

```python
my_documented_function.__doc__
```

More information about the conventions and standards around docstrings can be found - https://www.python.org/dev/peps/pep-0257/
